"bought_in_last_month contained web scraping 
artifacts like 'List:', 'ESRB', 'Typical:' 
and price strings like '($1,999.75/100)'.
These were invalid entries converted to NaN
during cleaning."

```python
#converting
df['bought_in_last_month']=df['bought_in_last_month'].str.split(' ').str[0].str.replace("+",'')
def clean_bought(x):
    if pd.isna(x):           # handle NaN
        return None
    x = str(x).strip()
    if 'K' in x:             # handle K values
        try:
            return float(x.replace('K','')) * 1000
        except:
            return None
    else:
        try:
            return float(x)  # handle numeric strings
        except:
            return None      # handle garbage like 'List', 'ESRB'

df['bought_in_last_month'] = df['bought_in_last_month'].apply(clean_bought)
```

Before deleting rows,i want to check whether the data distribution changes before and after dropping.It helps me to know whether the data distribution is significantly changing.

```python
#traditional way-Plot histogram to check the significant change of data distributions before and after deleting the rows
import matplotlib.pyplot as plt
plt.hist(df['rating'],bins=20)
plt.show()
plt.hist(temp_df['rating'],bins=20)
plt.show()
```

Using KDE plot(Kernel Density Estimation)-It is a smooth version of histogram.Instead of showing the bars,it shows a curve representing how data is distributed.It can be used to visualize the data distribution before and after deleting.
If two curves are overlapping that means deleting those missing values will not the data or creates problem while building model.Else it will creates problem while modeling.
in this plot,density tells highest concentration of ratings.

```python
sns.kdeplot(df['rating'], label='Before')
sns.kdeplot(temp_df['rating'], label='After')

plt.legend()
plt.show()
```

Following the same procedure for number_of_reviews

```python
sns.kdeplot(df['number_of_reviews'],label='before deleting')
sns.kdeplot(temp_df['number_of_reviews'],label='after deleting')
plt.legend()
plt.show()
```

Found that mean>median,so it would be best choice to impute with median as the distribution is heavily right skewed.It helps to avoid model improper analysis

```python
#filling the bought in last month with median using simpleimputer sklearn preprocessing tool
from sklearn.impute import SimpleImputer
imputer=SimpleImputer(strategy='median')
df['bought_in_last_month']=imputer.fit_transform(df[['bought_in_last_month']])
```
